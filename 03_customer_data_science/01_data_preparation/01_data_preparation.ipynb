# Customer Data Preparation

This notebook loads the customer-level analytical dataset from SQL Server
and performs initial data quality and structure checks before exploratory
data analysis and machine learning.

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import pyodbc
import os

from sqlalchemy import create_engine
from urllib.parse import quote_plus
from pathlib import Path
```

## 2. SQL Server Connection

```python
server = os.getenv("SQL_SERVER", "localhost")
database = "EcommerceAnalysis"

connection_string = (
    "DRIVER={ODBC Driver 18 for SQL Server};"
    f"SERVER={server};"
    f"DATABASE={database};"
    "Trusted_Connection=yes;"
    "TrustServerCertificate=yes;"
)
```

```python
encoded_connection_string = quote_plus(connection_string)
```

```python
engine = create_engine(
    f"mssql+pyodbc:///?odbc_connect={encoded_connection_string}"
)
```

## 3. Load Customer Dataset 

Load the customer-level analytical dataset prepared in SQL Server into a Pandas DataFrame.

```python
sql_path = Path("01_data_preparation") / "01_customer_dataset.sql"

sql_path
```

```python
customer_query = sql_path.read_text(encoding="utf-8")
```

```python
df = pd.read_sql(customer_query, engine)
```
